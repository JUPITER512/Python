```python
string2 = input("Input the string")
count = 0
repeatedCharacter = {}
for i in range(len(string2)-1):
    count = 0
    for j in range(len(string2)):
        if (string2[i] == string2[j]):
            count += 1
            repeatedCharacter[string2[i]] = count


{k: v for k, v in repeatedCharacter.items() if v > 1}
```
