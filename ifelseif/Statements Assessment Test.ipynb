```python
st = 'Print only the words that start with s in this sentence'
```

```python
wordlist=st.split()
```

```python
for word in wordlist:
    if(word.startswith('s')):
        print(word)
```

```python
st = 'Print every word in this sentence that has an even number of letter'
evenword=st.split()
```

```python
evenword
for word in evenword:
    if(len(word)%2==0):
        print(f'{word} even') 
```

```python
for i in range(1,100):
    if(i%3==0 and i%5==00):
        print('fizzBuzz')
    elif(i%3==0):
        print('fizz')
    elif(i%5==0):
        print('buzz')
    else:
        print(i)
```
