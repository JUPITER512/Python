```python
def old_macdonald(name):
    first=name[0].capitalize()
    between=name[1:3]
    fourth=name[3].capitalize()
    rest=name[4:]


    return first+between+fourth+rest
```

```python
old_macdonald('macdonald')
```

```python
def master_yoga(txt):
    reverse= txt.split()
    reverse=reverse[::-1]
    result= ' '.join(reverse)
    return result
```

```python
master_yoga('i am home')
```

```python
def almost_there(n):
    if(abs(100-n)<=10 or abs(200-n)<=10):
        return True
```

```python
almost_there(104)
```

```python
almost_there(150)
```
