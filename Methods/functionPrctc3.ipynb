```python
def has_33(number):
    for i in range(0, len(number)-1):
        if (number[i] == 3 and number[i+1] == 3):
            return True
        else:
            return False
```

```python
has_33([1, 3, 1, 3])
```

```python
def paper_doll(text):
    result=''
    for i in text:
        result=result+i*3

    return result
```

```python
paper_doll('Hello')
```

```python
def black_jack(a, b, c):
    sum = 0
    if (1 <= a <= 11 and 1 <= b <= 11 and 1 <= c <= 11):
        sum = a+b+c
        if (sum <= 21):
            return sum
        elif (sum > 21 and (11 in [a, b, c])):
            return sum-10
        else:
            return 'Bust'
```

```python

black_jack(5,6,7)
```

```python
black_jack(9,9,9)
```

```python
black_jack(9,9,11)
```

```python
def sumer_69(arr):
    total=0
    add=True
    for num in arr:
        while add:
            if num!=6:
                total+=num
                break
            else:
                add=False
        while not add:
            if num!=9:
                break
            else:
                add=True
                break
    return total
```

```python
sumer_69([4, 5, 6, 7, 8, 9])
```
