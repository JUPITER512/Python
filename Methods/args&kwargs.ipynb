```python
def myfun(a,b):
    return sum((a,b))*0.05
```

```python
myfun(40,50)
```

```python

def myfunc(*args):
    print(args)
    return sum(args)*0.05
```

```python
myfunc(40,60)
```

```python
myfunc(40,60,100,1,34)
```

```python
def myfunction(**kwargs):
    print(kwargs)
    if 'fruit' in kwargs:
        print ('My Fruit of choice is {} and fav veggie is {}' .format(kwargs['fruit'],kwargs['veggie']))
    else:
        print("Fruit not found")
```

```python
myfunction(fruit='apple',veggie='carrot')
```
