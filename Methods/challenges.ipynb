```python
def spy_game(num):
    code=[0,0,7,'x']
    
    for i in num:
        if i==code[0]:
            code.pop(0)
    return len(code)==1
    
```

```python
spy_game([1,2,4,0,0,7,5]) 
```

```python
spy_game([1,7,2,0,4,5,0])
```

```python
def find_sum_of_primes(minimum, maximum):
    total = 0
    for num in range(minimum, maximum + 1):
        is_prime = True
        if num <= 1:
            is_prime = False
        else:
            for i in range(2, int(num ** 0.5) + 1):
                if num % i == 0:
                    is_prime = False
                    break
        if is_prime:
            print(num, end=" ")
            total += num
    return total


minimum = int(input("Please enter the minimum value: "))
maximum = int(input("Please enter the maximum value: "))
prime_sum = find_sum_of_primes(minimum, maximum)

```
