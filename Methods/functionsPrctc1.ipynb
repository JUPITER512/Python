```python
def lesser_of_two_evens(a,b):
   
   if(a%2==0 and b%2==0):
        return min(a,b)
   elif(a%2!=0 or b%2!=0):
      return max(a,b)
    
```

```python
lesser_of_two_evens(2,4)
```

```python
lesser_of_two_evens (2,5)
```

```python
def animal_cracker(text):
    wordlist = text.lower().split()
    print(wordlist)
    return wordlist[0][0] == wordlist[1][0]
```

```python
animal_cracker('Crazy cat')
```

```python
def makes_twenty(n1,n2):
    sum=n1+n2
    if(sum==20 or n1==20 or n2==20):
        return True
    else:
        return False
```

```python
makes_twenty(20, 10)
```

```python
makes_twenty(2,3)
```
