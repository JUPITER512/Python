```python
from IPython.display import clear_output


def display_board(board):
    clear_output()
    print('   |   |')
    print(' '+board[7]+'  | '+board[8]+'  | '+board[9])
    print('   |   |')
    print('------------')
    print(' '+board[4]+'  | '+board[5]+'  | '+board[6])
    print('   |   |')
    print('------------')
    print(' '+board[1]+'  | '+board[2]+'  | '+board[3])
    print('   |   |')
```

```python
def player_input():
    '''
    Output=(Player 1 Marker, Player 2 Marker)

    '''

    marker=''

    while not (marker=='X' or marker=='O'):
        marker=input('Player1: Choose X or O:' ).upper()

    if marker=='X':
        return ('X','O')
    else:
        return('O','X')
```

```python
def place_marke(board,marker,position):
    board[position]=marker
```

```python
def win_check(board, mark):
    # win
    return((board[1] == mark and board[2] == mark and board[3] == mark) or
        (board[4] == mark and board[5] == mark and board[6] == mark) or
        (board[7] == mark and board[8] == mark and board[9] == mark) or
        (board[1] == mark and board[4] == mark and board[7] == mark) or
        (board[2] == mark and board[5] == mark and board[8] == mark) or
        (board[3] == mark and board[6] == mark and board[9] == mark) or
        (board[1] == mark and board[5] == mark and board[9] == mark) or
        (board[3] == mark and board[5] == mark and board[7] == mark))
```

```python
import random
def chose_first():
    flip=random.randint(0,1)
    if flip==0:
        return 'Player 1'
    else:
        return 'Player 2'
```

```python
def space_check(board,position):
    return board[position]==''
```

```python
def board_full(board):
    for i in range(1,10):
        if space_check(board,i):
            return False
    return True
```

```python
def player_choice(board):
    poisiton=0
    while(poisiton not in [1,2,3,4,5,6,7,8,9] or not space_check(board,poisiton)):
        poisiton=int(input('Input 1-9'))
    return poisiton
```

```python
def replay():
    choice=input("Play again>   ? Enter Yes or no ").lower()

    return choice=='yes'
```

```python
print("WELCOME TO THE TIC TAC TOE GAME")

while True:
    # play
    # set up (#board,who first, chose markers x,o)
    the_board = ['']*10
    player1_marker, player2_marker = player_input()
    turn = chose_first()
    print(turn+' Will Go First')
    play_game = input("Ready to Play? Yes or No").lower()
    if play_game.lower()[0] == 'y':
        game_on = True
    else:
        game_on = False

    # gameplay
    while game_on:
        if turn == 'Player 1':
            display_board(the_board)
            position = player_choice(the_board)
            place_marke(the_board, player1_marker, position)

            if win_check(the_board, player1_marker):
                display_board(the_board)
                print("PLAYER 1 won")
                game_on = False
            else:
                if board_full(the_board):
                    display_board(the_board)
                    print("Tie Game")
                    break
                else:
                    turn = 'Player 2'
        else:
            display_board(the_board)
            position=player_choice(the_board)
            place_marke(the_board,player2_marker,position)

            if win_check(the_board,player2_marker):
                display_board(the_board)
                print("PLAYER 2 won")
                game_on=False
            else:
                if board_full(the_board):
                    display_board(the_board)
                    print("Tie Game")
                    game_on=False
                else:
                    turn='Player 1'
    if not replay():
        break


    #player one turn
    #player two turn
```
