```python
class Dog():
    species='mamal'
    def __init__(self,breed,name):
        self.breed=breed
        self.name=name
    def bark(self,number):
        print(f'Woof {self.name} and number is {number}')
```

```python
my_dog=Dog(breed='lab',name='tyson')
```

```python
my_dog.breed
my_dog.name
my_dog.species
my_dog.bark(2)
```

```python
class Circle():
    pi=3.14
    def  __init__(self,radius=1):
        self.radius=radius
        self.area=(radius**2)*Circle.pi
    def getCircumfernce(self):
        return 2*Circle.pi*self.radius
    
```

```python
mycircle=Circle(30)

```

```python
mycircle.getCircumfernce()
```

```python
mycircle.pi
mycircle.radius
mycircle.area
```
