```python
class Book():
    def __init__(self,title,author,pages):
        self.title=title
        self.author=author
        self.pageas=pages

    def __str__(self):
        return f"{self.title} by {self.author}"
    def __len__(self):
        return self.pageas
    def __del__(self):
        print(" A bbook has been deleted")
```

```python
b=Book('Python rocks','Ali',200)
```

```python
print(b)
```

```python
len(b)
```
