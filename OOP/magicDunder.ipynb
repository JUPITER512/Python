```python
class Book():

    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"{self.author} wrote {self.title} having {self.pages}pages"

    def __len__(self):
        return self.pages

    def __del__(self):
        print("BOOK HAS BEEN DELETED")
```

```python
b = Book('Python Rocks', 'Ali Murtaza', 250)
```

```python
str(b)
```

```python
len(b)
```

```python
del b
```
