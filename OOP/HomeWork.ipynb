```python
import math
class Line():
    def __init__(self,cord1,cord2):
        self.cordinate1=cord1
        self.cordinate2=cord2
    def distance(self):
            x1,y1=self.cordinate1
            x2,y2=self.cordinate2
            distance=math.sqrt(((x2-x1)**2) + ((y2-y1)**2))
            return distance
    def slope(self):
            x1,y1=self.cordinate1
            x2,y2=self.cordinate2
            slope=(y2-y1)/(x2-x1)
            return slope
         
```

```python
cord1=(3,2)
cord2=(8,10)
myLine=Line(cord1,cord2)
```

```python
myLine.distance()
```

```python
myLine.slope()
```

```python
class Cylinder:
    pie=3.14


    def __init__(self,height=1,radius=1) :
        self.height=height
        self.radius=radius
    def volume(self):
        return (Cylinder.pie * ( self.radius**2)*self.height)
    def surfaceArea(self):
        return (2 * Cylinder.pie *self.height*self.radius)+(2*Cylinder.pie*(self.radius**2))

        
        
```

```python
mycyl=Cylinder(2,3)
```
