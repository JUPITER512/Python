```python
#inheritance
class Animal():
    def __init__(self):
        print("ANIMAL IS CREATED")
    def who_am_i(self):
        print("I AM An Animal ")
    def eat(self):
        print('I AM EATING')
```

```python
myanimal=Animal()
```

```python
myanimal.eat()
```

```python
myanimal.who_am_i()
```

```python
class Dog(Animal):
    def __init__(self):
        Animal.__init__(self)
        print("Dog Created")
    def who_am_i(self):
        print('I am a Dog ANimal')
    def eat(self):
        print("I AM DOG AND EATING  MEAT")
    def barkself(self):
        print("WOOF ")
```

```python
mydog=Dog()
```

```python
mydog.eat()
mydog.who_am_i()
mydog.barkself()
```

POLYMORPHISM

```python
class Dog():
    def __init__(self,name):
        self.name=name
    def speak(self):
        return self.name+" says woof"
```

```python
class Cat():
    def __init__(self,name):
        self.name=name
    def speak(self):
        return self.name+" says meow"
```

```python
flix=Dog('Flix')
```

```python
niko=Cat('niko')
```

```python
for i in [flix,niko]:
    print(i.speak())
```

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(flix)
```

```python
pet_speak(niko
)
```

Abstraction and Inheritance

```python
class Animal():
    def __init__(self,name):
        self.name=name
    def speak(self):
        raise NotImplementedError("Sub class must implement this methid")
```

```python
myanimal=Animal('fred')
```

```python
myanimal.speak()
```

```python
class Dog1(Animal):
    def speak(self):
        return self.name+' says woof'
```

```python
class Cat(Animal):
    def speak(self):
        return self.name+' says meow'
```

```python
fido=Dog1('FIDO')
```

```python
isis=Cat("isis")
```

```python
fido.speak()
```
