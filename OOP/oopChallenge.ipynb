```python
class BanK:

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self):
        amount=int(input("Please enter How much ammont do you want to deposit"))
        self.balance += amount
        print(f"This is {self.owner} current Balance now {self.balance}")
    def withdraw(self):
        amount=int(input("Please enter How much ammont do you want to Withdraw"))
        while(amount>self.balance):
            print(f"{self.owner} can not withdraw more amount than your current Balance\n")
            print("Re-enter valid ammoun")
            amount=int(input("Please enter How much ammont do you want to Withdraw"))
        print(f"{self.owner} Withdraw {amount} from your balance {self.balance}")
        self.balance-=amount
        print(f"{self.owner} current balance is now {self.balance}")
```

```python
mybank=BanK('Syed Ali',5000)
```

```python
mybank.deposit()
mybank.withdraw()
```
